# Анализ и визуализация данных по университетам

В этом ноутбуке строим графики по данным: universities.csv и university_students.csv.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import display

sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (11, 6)
```

## 1. Настройка ссылок на CSV

```python
base_url = "https://raw.githubusercontent.com/kirahilimanuck4921-cmyk/python-ai-Khilimanuck-Kirill/main/data"
universities_url = f"{base_url}/universities.csv"
students_url = f"{base_url}/university_students.csv"
```

## 2. Загрузка данных

```python
df_universities = pd.read_csv(universities_url)
df_students = pd.read_csv(students_url)

display(df_universities.head())
display(df_students.head())
```

## 3. Подготовка данных

```python
df_universities = df_universities.rename(columns={
    "universityLabel":"university",
    "countryLabel":"country",
    "cityLabel":"city",
    "foundingYear":"founding_year"
})

df_students = df_students.rename(columns={
    "universityLabel":"university",
    "totalStudents":"total_students",
    "undergradStudents":"undergrad_students",
    "postgradStudents":"postgrad_students"
})

df_universities["founding_year"] = pd.to_numeric(df_universities["founding_year"], errors="coerce")
df_students["total_students"] = pd.to_numeric(df_students["total_students"], errors="coerce")
df_students["undergrad_students"] = pd.to_numeric(df_students["undergrad_students"], errors="coerce")
df_students["postgrad_students"] = pd.to_numeric(df_students["postgrad_students"], errors="coerce")
```

## График 1. Топ-10 стран по количеству университетов

```python
top_countries = df_universities["country"].value_counts().head(10).reset_index()
top_countries.columns = ["country", "count"]

plt.figure(figsize=(11, 6))
sns.barplot(data=top_countries, x="count", y="country")
plt.title("Топ-10 стран по количеству университетов")
plt.xlabel("Количество университетов")
plt.ylabel("Страна")
plt.tight_layout()
plt.show()
```

## График 2. Распределение годов основания университетов

```python
years = df_universities["founding_year"].dropna()
plt.figure(figsize=(12, 6))
sns.histplot(years, bins=30, kde=True)
plt.title("Распределение годов основания университетов")
plt.xlabel("Год основания")
plt.ylabel("Количество университетов")
plt.tight_layout()
plt.show()
```

## График 3. Топ-10 университетов по общему числу студентов

```python
top_universities = df_students.sort_values("total_students", ascending=False).head(10)

plt.figure(figsize=(11, 6))
sns.barplot(data=top_universities, x="total_students", y="university")
plt.title("Топ-10 университетов по общему числу студентов")
plt.xlabel("Количество студентов")
plt.ylabel("Университет")
plt.tight_layout()
plt.show()
```

## График 4. Число бакалавров и магистрантов для топ-10 университетов

```python
top_universities_melt = top_universities.melt(id_vars="university", value_vars=["undergrad_students","postgrad_students"], var_name="type", value_name="count")

plt.figure(figsize=(12, 6))
sns.barplot(data=top_universities_melt, x="university", y="count", hue="type")
plt.title("Бакалавры и магистранты в топ-10 университетах")
plt.xlabel("Университет")
plt.ylabel("Количество студентов")
plt.xticks(rotation=45, ha="right")
plt.legend(title="Тип студентов")
plt.tight_layout()
plt.show()
```
