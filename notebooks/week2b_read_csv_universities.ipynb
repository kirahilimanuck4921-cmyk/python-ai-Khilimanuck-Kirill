# Первичная обработка данных по университетам

В этом ноутбуке загружаем два CSV-файла: universities.csv и university_students.csv, проверяем данные и делаем базовый анализ.

```python
import pandas as pd
from IPython.display import display
```

## 1. Настройка ссылок на CSV

```python
base_url = "https://raw.githubusercontent.com/kirahilimanuck4921-cmyk/python-ai-Khilimanuck-Kirill/main/data"
universities_url = f"{base_url}/universities.csv"
students_url = f"{base_url}/university_students.csv"
```

## 2. Загрузка CSV

```python
df_universities = pd.read_csv(universities_url)
df_students = pd.read_csv(students_url)

print(f"universities.csv: {df_universities.shape[0]} строк, {df_universities.shape[1]} столбцов")
print(f"university_students.csv: {df_students.shape[0]} строк, {df_students.shape[1]} столбцов")

display(df_universities.head())
display(df_students.head())
```

## 3. Проверка и переименование столбцов

```python
df_universities = df_universities.rename(columns={
    "universityLabel":"university",
    "countryLabel":"country",
    "cityLabel":"city",
    "foundingYear":"founding_year"
})

df_students = df_students.rename(columns={
    "universityLabel":"university",
    "totalStudents":"total_students",
    "undergradStudents":"undergrad_students",
    "postgradStudents":"postgrad_students"
})

display(df_universities.head())
display(df_students.head())
```

## 4. Преобразование типов данных

```python
df_universities["founding_year"] = pd.to_numeric(df_universities["founding_year"], errors="coerce")
df_students["total_students"] = pd.to_numeric(df_students["total_students"], errors="coerce")
df_students["undergrad_students"] = pd.to_numeric(df_students["undergrad_students"], errors="coerce")
df_students["postgrad_students"] = pd.to_numeric(df_students["postgrad_students"], errors="coerce")
```

## 5. Проверка пропусков

```python
print("Пропуски в universities.csv:")
display(df_universities.isna().sum())

print("\nПропуски в university_students.csv:")
display(df_students.isna().sum())
```

## 6. Базовая статистика

```python
display(df_universities.describe(include='all'))
display(df_students.describe(include='all'))
```
